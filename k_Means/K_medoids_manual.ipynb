```python
import numpy as np
```

```python
X = np.array([
    [1,1],
    [2,2],
    [3,1],
    [8,8],
    [9,9],
    [8,9]
])
```

```python
names = ['A','B','C','D','E','F']
```

```python
K =2
```

```python
medoid_indices = [0,3]
```

```python
print("Initial Medoids:")
for i in medoid_indices:
  print(names[i],"->",X[i])
```

```python
def manhattan_distance(point1, point2):
  return np.sum(np.abs(point1 - point2))
```

```python
distances = np.zeros((len(X),K))
```

```python
for i in range(len(X)):
  for j in range(K):
    distances[i][j] = manhattan_distance(
        X[i],
        X[medoid_indices[j]]
    )
```

```python
print("\nmanhattan Distances:")
print(distances)
```

```python
clusters = np.argmin(distances, axis=1)
```

```python
print("\nCluster Assignment:")

for i in range(len(X)):
  print(
      names[i],
      "-> Cluster",
      clusters[i] + 1
  )
```

```python
total_cost = 0
```

```python
for i in range(len(X)):
  total_cost += distances[i][clusters[i]]

print("\n Total Cost", total_cost)
```

```python
new_medoid_indices = []

for cluster in range(K):

  cluster_points = np.where(clusters == cluster)[0]

  best_medoid = cluster_points[0]
  best_cost = float('inf')

  for candidate in cluster_points:

    cost = 0

    for point in cluster_points:
      cost += manhattan_distance(X[point],X[candidate])

    if cost < best_cost:
      best_cost = cost
      best_medoid = candidate

  new_medoid_indices.append(best_medoid)
```

```python
print("\nNew Medoids:")

for i in new_medoid_indices:
  print(names[i],"->",X[i])
```
