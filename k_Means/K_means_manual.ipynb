```python
import numpy as np
```

```python
X = np.array([
    [1,1],
    [2,2],
    [3,1],
    [8,8],
    [9,9],
    [8,9]

])
```

```python
names = ['A','B','C','D','E','F']
```

```python
K = 2
```

```python
centroids = np.array([
    [1,1],
    [8,8]
], dtype=float)
```

```python
print("Intial Centroids")
print(centroids)
```

```python
for iteration in range(10):

  print("\nIteration ", iteration +1)

  distances = np.sqrt(
      ((X[:, np.newaxis, :] - centroids[np.newaxis,:,:]) ** 2).sum(axis=2)
  )

  print("\n Distances:")
  print(distances)

  clusters = np.argmin(distances, axis=1)

  print("\nCluster Assignment:")
  for i in range(len(X)):
    print(names[i],"-> Cluster", clusters[i] +1)

  new_centroids = np.array([
      X[clusters == k].mean(axis=0)
      for k in range(K)
  ])

  print("\nNew Centroids:")
  print(new_centroids)

  if np.allclose(centroids, new_centroids):
    print("\nCentroids have stopped changing.")
    break

  centroids = new_centroids
```

```python
print("\nFinal Centroids: ")
print(centroids)

print("\nFinal Clusters:")

for k in range(K):
  print("\nCluster", k+1)

  for i in range(len(X)):
    if clusters[i] == k:
      print(names[i], "->", X[i])
```
