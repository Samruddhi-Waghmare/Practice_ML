```python
import numpy as np
```

```python
X = np.array([
    [2,6],
    [4,5],
    [6,7],
    [8,8],
    [10,9]
])
```

```python
y = np.array([50,60,65,80,90])
```

```python
X_b = np.c_[np.ones((X.shape[0],1)),X]
```

```python
beta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y
```

```python
print("Intercept:",beta[0])
print("Coefficient for x1",beta[1])
print("Coefficient for x2", beta[2])
```

```python
new_data = np.array([1,7,8])
prediction = new_data @ beta
print("Predicted Score", prediction)
```
