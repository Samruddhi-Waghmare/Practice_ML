```python
import numpy as np
```

```python
states = ["Rainy","Sunny"]
observations = ['Walk','Shop','Clean']
```

```python
obs_seq = [0,1,2]
```

```python
pi = np.array([0.6,0.4])
```

```python
A = np.array([
    [0.7, 0.3],
    [0.4,0.6]
])
```

```python
B = np.array([
    [0.1,0.4,0.5],
    [0.6,0.3,0.1]
])
```

```python
def forward(obs_seq, A,B, pi):

  N = len(A)

  T = len(obs_seq)

  alpha = np.zeros((T,N))

  alpha[0] = pi * B[:, obs_seq[0]]

  for t in range(1,T):

    for j in range(N):

      alpha[t,j] = (
          np.sum(alpha[t-1] * A[:,j])*B[j, obs_seq[t]]
      )

  return np.sum(alpha[T-1])
```

```python
forward_prob = forward(obs_seq, A,B,pi)
```

```python
print("Forward Algorithm:")
print("Probability of Observations =", forward_prob)

print()
```

```python
def viterbi(obs_seq, A, B, pi):

  N = len(A)

  T = len(obs_seq)

  delta = np.zeros((T,N))

  psi = np.zeros((T,N), dtype = int)

  delta[0] = pi * B[:, obs_seq[0]]

  for t in range(1, T):

    for j in range(N):

      seq_probs = delta[t-1] * A[:, j]

      psi[t,j] = np.argmax(seq_probs)

      delta[t,j] = (
          np.max(seq_probs) * B[j, obs_seq[t]]
      )


  best_path_prob = np.max(delta[T-1])

  best_last_state = np.argmax(delta[T-1])

  best_path = [best_last_state]

  for t in range(T-1,0, -1):

    best_last_state = psi[t, best_last_state]

    best_path.insert(0, best_last_state)

  best_path = [states[i] for i in best_path]

  return best_path_prob, best_path
```

```python
viterbi_prob, viterbi_path = viterbi(
    obs_seq, A,B, pi
)
```

```python
print("Viterbi Algorithm")
print("Best Path Probability ", viterbi_prob)
print("Most Likely State Sequence", viterbi_path)
```
