```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([
   [1],
   [2],
   [3],
   [4],
   [5],
   [6],
   [7],
   [8]

])
```

```python
y = np.array([
    0,0,0,0,1,1,1,1
])
```

```python
def sigmoid(z):
  return 1/(1+np.exp(-z))
```

```python
w = 0
b = 0
learning_rate = 0.1
epochs = 1000
```

```python
n = len(X)
```

```python
for i in range(epochs):
  z = w * X.flatten() + b
  y_pred = sigmoid(z)
  dw =  (1/n) * np.sum((y_pred - y) * X.flatten())
  db = (1/n) * np.sum(y_pred - y)
  w = w - learning_rate * dw
  b = b - learning_rate * db
```

```python
w = w - learning_rate * dw
b = b - learning_rate * db
```

```python
def predict(X):
  z = w * X + b
  probability = sigmoid(z)
  if probability >= 0.5:
    return 1
  else:
    return 0
```

```python
test_values = np.array([2,4,5,7])

for value in test_values:
  probability = sigmoid(w * value + b)
  prediction = 1 if probability >= 0.5 else 0
  print (
    "Input:",value,
    "probability:",probability,
    "prediction:",prediction
  )
```
