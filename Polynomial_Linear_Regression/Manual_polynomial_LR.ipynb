```python
import numpy as np
```

```python
X = np.array(
    [1,2,3,4,5,]
)
```

```python
y = np.array([35,45,60,82,110])
```

```python
X_poly = np.column_stack((np.ones(len(X)),X, X**2))
```

```python
print("Polynomial Features")
print(X_poly)
```

```python
beta = np.linalg.inv(X_poly.T @ X_poly) @ X_poly.T @ y

print("Intercept:", beta[0])
print("Coefficient of X:", beta[1])
print("Coefficient of X²:", beta[2])
```

```python
x_new = np.array([1, 6, 36])   # [1, X, X²]
prediction = x_new @ beta

print("Predicted Value:", prediction)
```
