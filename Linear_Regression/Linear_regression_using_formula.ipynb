```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("cgpa.csv")
```

```python
X = df['cgpa']
y=df['package']
```

NAME OF DATA POINTS

```python
n = len(X)
```

CALCULATE SLOPE = m

```python
m = (n * np.sum(X * y) - np.sum(X) * np.sum(y)) / (n * np.sum(X ** 2) - np.sum(X) ** 2)
```

CALCULATE INTERCEPT = C

```python
c = (np.sum(y)-m*np.sum(X))/n
```

```python
print("slope",m)
print("intercept",c)
```

FORMULA OF REGRESSION LINE

```python
y_pred = m*X +c
```

PLOT OF THE PREDICTION

```python
plt.scatter(X,y,label="Actual data")
plt.plot(X,y_pred,label="Regression Line", color='red')
plt.xlabel("cgpa")
plt.ylabel("package")
plt.title("Linear Regression using Formula")
plt.legend()
plt.show()
```
